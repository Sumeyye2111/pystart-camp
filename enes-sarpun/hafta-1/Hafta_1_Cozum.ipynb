### Soru 1: Katılımcı Kartı

Kullanıcıdan adını, soyadını, bölümünü ve doğum tarihini (gün, ay ve yıl ayrı ayrı) `input()` ile al, aşağıdaki gibi bir kart yazdır.

- Doğum tarihini `(gün, ay, yıl)` şeklinde bir **tuple**'da tut.
- Yaşı `2026 - doğum yılı` olarak hesapla.
- Ad soyadın baş harflerini büyüt, tarihi iki haneli yazdır (`07.03.2006`).

> **İpucu:** `f"{gun:02d}"` sayıyı iki haneye tamamlar.

**Örnek Çıktı Biçimi:**
```text
========================================
PyStart Katılımcı Kartı
========================================
Ad Soyad    : Ayşe Yılmaz
Bölüm       : Bilgisayar Mühendisliği
Doğum       : 07.03.2006
Yaş         : 20
========================================
```

````python
# --- Soru 1 Kod Alanı ---
ad = input("Adın: ")
soyad = input("Soyadın: ")
bolum = input("Bölümün: ")
gun = int(input("Doğum günü: "))
ay = int(input("Doğum ayı: "))
yil = int(input("Doğum yılı: "))

# Doğum tarihi bir tuple: (gün, ay, yıl). Değişkene "tuple" adını vermiyoruz,
# çünkü bu Python'un kendi tuple() fonksiyonunun adı.
dogum_tarihi = (gun, ay, yil)

yas = 2026 - dogum_tarihi[2]

# title() her kelimenin baş harfini büyütür.
# Dikkat: "ilke".title() -> "Ilke" olur; Türkçe İ/I sorunu burada da var (Soru 2'ye bak).
ad_soyad = f"{ad} {soyad}".title()

print("=" * 40)
print("PyStart Katılımcı Kartı")
print("=" * 40)
print(f"Ad Soyad    : {ad_soyad}")
print(f"Bölüm       : {bolum}")
print(f"Doğum       : {dogum_tarihi[0]:02d}.{dogum_tarihi[1]:02d}.{dogum_tarihi[2]}")
print(f"Yaş         : {yas}")
print("=" * 40)
````

### Soru 2: Cümle Analizi

Kullanıcıdan bir cümle al ve şunları yazdır:

1. Karakter sayısı ve boşluklar hariç karakter sayısı
2. Kelime sayısı, ilk kelime ve son kelime
3. Cümlenin tamamen büyük harfli hali ve tersten yazılışı
4. Cümlede kaç tane 'a' harfi olduğu (büyük-küçük fark etmeden)
5. Cümlede 'python' kelimesinin geçip geçmediği (`True` ya da `False`)

> **Gözlem:** İçinde 'i' harfi olan bir cümlenin büyük harfli haline dikkatle bak. Ne fark ettin? Gözlemini yorum satırı olarak yaz.

````python
# --- Soru 2 Kod Alanı ---
cumle = input("Bir cümle yaz: ")
kelimeler = cumle.split()   # boşluklardan bölüp kelime listesi yapar

# 1. Karakter sayıları
print("Karakter sayısı:", len(cumle))
print("Boşluklar hariç karakter sayısı:", len(cumle.replace(" ", "")))

# 2. Kelime sayısı, ilk ve son kelime
print("Kelime sayısı:", len(kelimeler))
print("İlk kelime:", kelimeler[0])
print("Son kelime:", kelimeler[-1])

# 3. Büyük harfli hali ve tersten yazılışı (ikisi ayrı ayrı)
print("Büyük harfli hali:", cumle.upper())
print("Tersten yazılışı:", cumle[::-1])

# 4. 'a' sayısı: önce hepsini küçültürsek 'A'lar da sayılır
print("'a' harfi sayısı:", cumle.lower().count("a"))

# 5. 'python' geçiyor mu? lower() olmadan "Python" yakalanmaz.
#    Aranan metin solda, cümle sağda: "python" in cumle (tersi yanlış olur)
print("'python' geçiyor mu:", "python" in cumle.lower())

# Gözlem Yorumunuz:
# upper() Türkçe kurallarını bilmiyor: küçük 'i' büyütülünce 'İ' değil noktasız 'I' oluyor.
# Örneğin "istanbul".upper() -> "ISTANBUL". Tersi de sorunlu: "İ".lower() tek bir harf değil,
# 'i' + görünmeyen bir nokta işareti olur, bu yüzden len("İ".lower()) 2 çıkar.
# Türkçe metinlerde upper(), lower() ve title() kullanırken buna dikkat etmek gerekir.
````

### Soru 3: Not Listesi

Kullanıcıdan üç notu tek satırda, aralarında boşlukla al (`70 85 90`). Notları sayıya çevirip bir listeye koy, ardından:

1. Kullanıcıdan bir not daha alıp listeye ekle.
2. Listeyi küçükten büyüğe sırala ve yazdır.
3. En düşük notu, en yüksek notu ve ortalamayı yazdır.
4. En düşük notu listeden çıkar ve yeni ortalamayı yazdır.

> **İpucu:** `split()` metni boşluklardan böler ama parçalar hâlâ metindir.

**Örnek Çıktı:**
```text
3 notu aralarında boşluk bırakarak yaz: 70 85 90
Bir not daha ekle: 60
Notlar (sıralı): [60, 70, 85, 90]
En düşük: 60 | En yüksek: 90
Ortalama: 76.25
En düşük not çıkarıldı: [70, 85, 90]
Yeni ortalama: 81.67
```

````python
# --- Soru 3 Kod Alanı ---
notlar_metin = input("3 notu aralarında boşluk bırakarak yaz: ")

# split() parçaları hâlâ metin ('70', '85', '90'); int() ile sayıya çeviriyoruz.
# Çevirmezsek sort/min/max metin karşılaştırır, sum hata verir.
parcalar = notlar_metin.split()
notlar = [int(parcalar[0]), int(parcalar[1]), int(parcalar[2])]

# 1. Bir not daha ekle
yeni_not = int(input("Bir not daha ekle: "))
notlar.append(yeni_not)

# 2. Küçükten büyüğe sırala
notlar.sort()
print("Notlar (sıralı):", notlar)

# 3. En düşük, en yüksek, ortalama (sabit sayı değil, listeden hesaplıyoruz)
print("En düşük:", min(notlar), "| En yüksek:", max(notlar))
print(f"Ortalama: {sum(notlar) / len(notlar):.2f}")

# 4. En düşüğü çıkar ve ortalamayı YENİDEN hesapla
notlar.remove(min(notlar))
print("En düşük not çıkarıldı:", notlar)
print(f"Yeni ortalama: {sum(notlar) / len(notlar):.2f}")
````

### Soru 4: Telefon Rehberi

Şu sözlükle başla:
```python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}
```

1. Kullanıcıdan bir isim ve numara alıp rehbere ekle.
2. Ali'nin numarasını `'0555 999 88 77'` olarak güncelle.
3. Ayşe'yi rehberden sil ve silinen numarayı yazdır.
4. Kullanıcıdan bir isim al ve numarasını yazdır. İsim rehberde yoksa program hata vermesin, `'Kayıt bulunamadı'` yazsın.
5. Rehberdeki kişi sayısını ve isimleri alfabetik sırayla yazdır.

> **İpucu:** Sözlüklerin `get()` metoduna ikinci bir varsayılan değer verebilirsin.

````python
rehber = {'Ali': '0555 111 22 33', 'Ayşe': '0532 444 55 66'}

# --- Soru 4 Kod Alanı ---
# Rehberi yeniden tanımlamıyoruz; yukarıdaki sözlükle devam ediyoruz.

# 1. Kullanıcıdan isim ve numara alıp ekle (örnek numara kullan, kendi numaranı yazma)
isim = input("Eklenecek kişinin adı: ")
numara = input("Numarası: ")
rehber[isim] = numara

# 2. Ali'nin numarasını güncelle
rehber['Ali'] = '0555 999 88 77'

# 3. Ayşe'yi sil: pop() silinen değeri zaten döndürür
silinen = rehber.pop('Ayşe')
print("Silinen numara:", silinen)

# 4. İsim ara; yoksa hata yerine varsayılan değer dönsün
aranan = input("Aranacak isim: ")
print("Numara:", rehber.get(aranan, 'Kayıt bulunamadı'))

# 5. Kişi sayısı ve alfabetik isimler (sorted sözlüğün anahtarlarını sıralar)
print("Kişi sayısı:", len(rehber))
print("İsimler (alfabetik):", sorted(rehber))
````

### Soru 5: Ortak Hobiler

İki kişinin hobilerini virgülle ayrılmış olarak al (`futbol,satranç,müzik`) ve her birini bir kümeye (`set`) çevir. Şunları yazdır:

1. Ortak hobiler, yalnızca birinci kişide olanlar, yalnızca ikinci kişide olanlar ve tüm hobiler
2. Şu soruların cevabı (`True` ya da `False` olarak):
   - Ortak hobiniz var mı?
   - Hobi sayılarınız eşit mi?
   - İkiniz de en az 3 hobi yazdınız mı?
   - 'satranç' ikinizden birinin hobisi mi?

> **İpucu:** Kümeleri yazdırırken sıra her çalıştırmada değişebilir; `sorted()` ile sabitleyebilirsin.

````python
# --- Soru 5 Kod Alanı ---
hobiler1 = input("1. kişinin hobileri (virgülle ayır): ")
hobiler2 = input("2. kişinin hobileri (virgülle ayır): ")

# Virgülden sonra boşluk bırakılırsa split(",") ' satranç' gibi başında boşluk olan
# parçalar üretir ve 'satranç' ile eşleşmez. replace(", ", ",") bunu temizler,
# lower() da 'Futbol' ile 'futbol'u aynı yapar.
kume1 = set(hobiler1.lower().replace(", ", ",").split(","))
kume2 = set(hobiler2.lower().replace(", ", ",").split(","))

# 1. Küme işlemleri (sorted ile sıralı ve her çalıştırmada aynı)
print("Ortak hobiler:", sorted(kume1 & kume2))
print("Yalnızca 1. kişide olanlar:", sorted(kume1 - kume2))
print("Yalnızca 2. kişide olanlar:", sorted(kume2 - kume1))
print("Tüm hobiler:", sorted(kume1 | kume2))

# 2. True / False soruları
# and / or ile bağlarken her karşılaştırmayı tam yaz:
#   len(kume1) >= 3 and len(kume2) >= 3   (doğru)
#   len(kume1) and len(kume2) >= 3        (yanlış: ilki sadece "boş mu" diye bakar)
#   len(kume1) >= 3 & len(kume2) >= 3     (yanlış: & önce çalışır)
print("Ortak hobiniz var mı?", len(kume1 & kume2) > 0)
print("Hobi sayılarınız eşit mi?", len(kume1) == len(kume2))
print("İkiniz de en az 3 hobi yazdınız mı?", len(kume1) >= 3 and len(kume2) >= 3)
print("'satranç' ikinizden birinin hobisi mi?", "satranç" in kume1 or "satranç" in kume2)
````
